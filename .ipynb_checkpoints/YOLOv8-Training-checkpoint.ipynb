### 1.3 Project Configuration & Directory Setup

All file paths are defined here. Only `PROJECT_ROOT`, `RUN_NAME`, and `TEST_IMAGE_PATH` should ever need to change. **Run this cell every time you open the notebook.**

```python
# Project Root (adjust to your local setup)
PROJECT_ROOT = os.path.expanduser("~/Documents/Thesis")

# Core Directories
DATA_DIR   = os.path.join(PROJECT_ROOT, "data")
RUNS_DIR   = os.path.join(PROJECT_ROOT, "runs")
MODELS_DIR = os.path.join(PROJECT_ROOT, "models")

# Dataset Paths (XD38M DATASET)
DATASET_DIR       = os.path.join(DATA_DIR, "coral-detection-1")
FIXED_DATASET_DIR = os.path.join(DATA_DIR, "coral-detection-1-fixed")
YAML_PATH         = os.path.join(FIXED_DATASET_DIR, "data.yaml")

# Dataset Paths (CORAL-QUADRAT-2 DATASET)
CORAL_QUAD_DIR = os.path.join(DATA_DIR, "coral-quadrat-2")
FXD_CORAL_QUAD_DIR = os.path.join(DATA_DIR, "coral-quadrat-2-fixed")
FNL_CORAL_QUAD_DIR = os.path.join(DATA_DIR, "coral-quadrat-2-final")
CORAL_QUAD_YAML_PATH = os.path.join(FNL_CORAL_QUAD_DIR, "data.yaml")

# Training Run Name
RUN_NAME = "Test_Run_V7"  # Change this for each new run

# Derived Paths
BEST_WEIGHTS_PATH = os.path.join(RUNS_DIR, RUN_NAME, "weights", "best.pt")
VERIFY_DIR        = os.path.join(DATA_DIR, "verify_output")
VERIFY_CORAL_QUAD_DIR   = os.path.join(DATA_DIR, "verify_coral-quad_output")
TEST_IMAGE_PATH   = os.path.join(PROJECT_ROOT, "test_images", "GOPR3275.jpg")  # ← change image as needed
CORALNET_TEST_DIR = Path(PROJECT_ROOT, "test_images")

# Create Directories
for d in [DATA_DIR, RUNS_DIR, MODELS_DIR, VERIFY_DIR]:
    os.makedirs(d, exist_ok=True)

print(f"Project root  : {PROJECT_ROOT}")
print(f"Data directory: {DATA_DIR}")
print(f"Runs directory: {RUNS_DIR}")
print(f"Active run    : {RUN_NAME}")
```

## Section 5: Model Initialization

Downloads and initializes the **YOLOv8-small** base model (`yolov8s.pt`). If the weights already exist locally, Ultralytics will reuse the cached file. **Run this cell every time before training.**

```python
print("Loading YOLOv8-small base model...")
model = YOLO("yolov8s.pt")
print("Model loaded successfully.")
```

---
## Section 6: Model Training

Trains YOLOv8 on the reconfigured single-class dataset (`coral-detection-1-fixed`).

> **Hardware note:** Batch size is set to `16` for an NVIDIA GeForce RTX 4060 Ti (8 GB VRAM). Reduce to `4` or `8` if you encounter CUDA out-of-memory errors.

> **Epochs & patience** were increased from 100 → 200 and 10 → 50 respectively, to prevent premature early stopping during plateau phases.

```python
print(f"Training YOLOv8 on: {YAML_PATH}\n")

results = model.train(
    data=YAML_PATH,

    # Training schedule
    epochs=200,
    imgsz=640,
    batch=16,
    patience=50,

    # Hardware
    device=0,

    # Optimizer
    optimizer="AdamW",
    lr0=0.001,
    lrf=0.01,
    momentum=0.937,
    weight_decay=0.0005,

    # Loss weights
    box=7.5,
    cls=0.5,
    dfl=1.5,

    # Augmentations
    hsv_h=0.02,
    hsv_s=0.8,
    hsv_v=0.5,
    translate=0.1,
    scale=0.6,
    fliplr=0.5,
    flipud=0.5,
    mosaic=1.0,
    mixup=0.1,
    cutmix=0.0,
    erasing=0.2,
    auto_augment="randaugment",

    # Output
    project=RUNS_DIR,
    name=RUN_NAME,
    verbose=True,
    plots=True,
)

print("\nTraining complete!")
```

---
## Section 8: Inference on CoralNet Images

### 8.1 Load Best Weights

Loads the best checkpoint from the completed training run. Update `RUN_NAME` in Section 1.3 if you need to load from a different run.

```python
print("Best weights path:", BEST_WEIGHTS_PATH)
model = YOLO(BEST_WEIGHTS_PATH)
print("Model loaded.")
```

### 8.2 Standard Inference

Run inference on a single CoralNet image. Update `TEST_IMAGE_PATH` in Section 1.3 to change the target image.

```python
print("Image path:", TEST_IMAGE_PATH)

results = model.predict(
    source=TEST_IMAGE_PATH,
    conf=0.01,           # Very low threshold — reveals all candidate detections
    iou=0.3,             # Loose NMS
    imgsz=1280,          # Higher resolution for CoralNet imagery
    save=True,
    save_txt=True,       # Save box coordinates and confidence scores
    project=RUNS_DIR,
    name="coralnet_test",
)

print("Inference complete.")
```

### 8.3 Check Detection Scores

Inspect how many detections were found and their confidence distribution. Useful for diagnosing whether predictions are present but suppressed vs. truly absent.

```python
for r in results:
    print(f"Detections at conf=0.01 : {len(r.boxes)}")
    if len(r.boxes) > 0:
        print("Confidence scores       :", r.boxes.conf.tolist())
```

### 8.4 SAHI Tiled Inference

Uses slice-and-predict to improve detection of small coral colonies in high-resolution images. Each 640×640 px tile overlaps its neighbors by 20% horizontally and vertically.

```python
detection_model = AutoDetectionModel.from_pretrained(
    model_type="yolov8",
    model_path=BEST_WEIGHTS_PATH,
    confidence_threshold=0.1, # Adviser's recommendation
    device="cuda:0", # Use GPU if available
)

result = get_sliced_prediction(
    TEST_IMAGE_PATH,
    detection_model,
    slice_height=640,
    slice_width=640,
    overlap_height_ratio=0.2,
    overlap_width_ratio=0.2,
)

sahi_output_dir = os.path.join(RUNS_DIR, "sahi_predictions")
os.makedirs(sahi_output_dir, exist_ok=True)
result.export_visuals(export_dir=sahi_output_dir)

print(f"Total detections : {len(result.object_prediction_list)}")
print(f"Results saved to : {sahi_output_dir}")
```

---
## Section 9: Training Curve Monitoring

Plots box loss, mAP@0.5, and recall curves from the training run's `results.csv`. Run after training is complete. The figure is also saved alongside the CSV in the run folder.

```python
results_csv_path = os.path.join(RUNS_DIR, RUN_NAME, "results.csv")
df = pd.read_csv(results_csv_path)
df.columns = df.columns.str.strip()

fig, axes = plt.subplots(2, 2, figsize=(12, 8))
df["train/box_loss"].plot(ax=axes[0, 0], title="Train Box Loss")
df["val/box_loss"].plot(ax=axes[0, 1], title="Val Box Loss")
df["metrics/mAP50(B)"].plot(ax=axes[1, 0], title="mAP@0.5")
df["metrics/recall(B)"].plot(ax=axes[1, 1], title="Recall")
plt.suptitle(f"Training Curves — {RUN_NAME}", fontsize=14)
plt.tight_layout()

curves_path = os.path.join(RUNS_DIR, RUN_NAME, "training_curves.png")
plt.savefig(curves_path)
plt.show()
print(f"Saved to: {curves_path}")
```
