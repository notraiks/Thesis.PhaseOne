### 2.3 EXIF-preserving visualization export

`result.export_visuals()` always writes PNG via `cv2.imwrite` under the hood
(confirmed by reading SAHI's own source: `sahi.utils.cv.visualize_object_predictions`) --
`cv2.imwrite` never carries EXIF, in any format, under any circumstances. Every
inference output this notebook produced was silently losing Make/Model/
DateTimeOriginal/MakerNote.

The helper below bypasses `export_visuals()` and calls SAHI's underlying
drawing function directly with `output_dir=None`, then saves the result
ourselves via Pillow with the source cropped image's EXIF re-attached --
same pattern as the crop script, verified round-trip before being used here.

```python
def export_visual_with_exif(result, source_image_path: Path, output_path: Path,
                             hide_labels: bool = False, hide_conf: bool = False):
    """
    Draws SAHI's detection boxes and saves the visualization as a JPEG that
    carries over source_image_path's EXIF metadata.

    result: a SAHI PredictionResult (has .image and .object_prediction_list)
    source_image_path: the CROPPED image that was actually run through inference
                        (not the raw image) -- its EXIF is what gets carried over
    output_path: full path to write the JPEG to (parent dirs created as needed)
    """
    draw_result = visualize_object_predictions(
        image=np.ascontiguousarray(result.image),
        object_prediction_list=result.object_prediction_list,
        hide_labels=hide_labels,
        hide_conf=hide_conf,
        output_dir=None,  # keep the drawn image in memory -- we save it ourselves
    )
    vis_image = Image.fromarray(draw_result["image"])

    with Image.open(source_image_path) as src_img:
        exif_bytes = src_img.info.get("exif")

    if exif_bytes:
        try:
            exif_dict = piexif.load(exif_bytes)
            exif_dict["Exif"][piexif.ExifIFD.PixelXDimension] = vis_image.width
            exif_dict["Exif"][piexif.ExifIFD.PixelYDimension] = vis_image.height
            exif_bytes = piexif.dump(exif_dict)
        except Exception:
            pass  # malformed/unreadable EXIF block -- write it through unchanged
            # rather than dropping it or failing the whole image, same fallback
            # behavior as crop_coralnet_quadrat.py.

    output_path.parent.mkdir(parents=True, exist_ok=True)
    save_kwargs = {"quality": 95}
    if exif_bytes:
        save_kwargs["exif"] = exif_bytes
    vis_image.save(output_path, **save_kwargs)


print("export_visual_with_exif ready -- used in place of result.export_visuals() below.")
```

### 2.4 SAHI Configuration Setup (*Finalized*)

`POSTPROCESS_MATCH_THRESHOLD = 0.5`, `VISUAL_DISPLAY_THRESHOLD = 0.15` — matches the
evaluation configuration that produced the visual results this decision was based
on. Not tuned further here.

```python
POSTPROCESS_TYPE = "GREEDYNMM"
POSTPROCESS_MATCH_METRIC = "IOS"
POSTPROCESS_MATCH_THRESHOLD = 0.5
VISUAL_DISPLAY_THRESHOLD = 0.15
SLICE_SIZE = 640
OVERLAP_RATIO = 0.2

detection_model = AutoDetectionModel.from_pretrained(
    model_type="ultralytics",
    model_path=str(WEIGHTS_PATH),
    confidence_threshold=VISUAL_DISPLAY_THRESHOLD,
    device="cuda:0",
)

SAHI_KWARGS = dict(
    slice_height=SLICE_SIZE, slice_width=SLICE_SIZE,
    overlap_height_ratio=OVERLAP_RATIO, overlap_width_ratio=OVERLAP_RATIO,
    postprocess_type=POSTPROCESS_TYPE,
    postprocess_match_metric=POSTPROCESS_MATCH_METRIC,
    postprocess_match_threshold=POSTPROCESS_MATCH_THRESHOLD,
)

_force_postprocess_supported = None

def sliced_predict_safe(image_path: str, **kwargs):
    """Same graceful-fallback pattern used throughout this project."""
    global _force_postprocess_supported
    if _force_postprocess_supported is not False:
        try:
            result = get_sliced_prediction(image_path, detection_model, force_postprocess_type=True, **kwargs)
            _force_postprocess_supported = True
            return result
        except TypeError:
            _force_postprocess_supported = False
    return get_sliced_prediction(image_path, detection_model, **kwargs)

print("SAHI configuration set. force_postprocess_type support confirmed on first call below.")
```

## SECTION 3: Inference Stage

### 3.1 CoralNet Dataset Full Inference

Optional — run when ready to process the complete corpus, not just the sample.
Resumable: images already processed (tracked in `processed_log.txt`) are skipped
on re-run, so this can safely be stopped and restarted if interrupted partway
through — worth having given this project has already lost training progress
once before.

```python
import csv as _csv

full_vis_dir = Path(FINAL_OUTPUT_DIR, "visuals_full")
full_vis_dir.mkdir(parents=True, exist_ok=True)

processed_log_path = Path(FINAL_OUTPUT_DIR, "processed_log.txt")
already_processed = set()
if processed_log_path.exists():
    already_processed = set(processed_log_path.read_text().splitlines())

detections_csv_path = Path(FINAL_OUTPUT_DIR, "full_dataset_detections.csv")
write_header = not detections_csv_path.exists()

with open(detections_csv_path, "a", newline="") as f, open(processed_log_path, "a") as log_f:
    writer = _csv.writer(f)
    if write_header:
        writer.writerow(["image_relpath", "num_detections"])

    remaining = [p for p in cropped_images if str(p.relative_to(OUTPUT_ROOT)) not in already_processed]
    print(f"{len(already_processed)} already processed, {len(remaining)} remaining.")

    for i, img_path in enumerate(remaining, 1):
        rel_path = img_path.relative_to(OUTPUT_ROOT)
        result = sliced_predict_safe(str(img_path), **SAHI_KWARGS)

        # Mirrors OUTPUT_ROOT's site/transect folder structure under
        # full_vis_dir -- same pattern crop_coralnet_quadrat.py already uses
        # for INPUT_ROOT -> OUTPUT_ROOT. Previously all ~900 visuals were
        # dumped flat into one folder with a hash appended to the filename
        # to avoid collisions across site folders; preserving the real
        # folder structure makes filenames unique on its own (no hash
        # needed) and keeps results browsable by site/transect.
        out_path = (full_vis_dir / rel_path).with_suffix(".jpg")
        export_visual_with_exif(result, img_path, out_path)

        n_detections = len(result.object_prediction_list)
        writer.writerow([str(rel_path), n_detections])
        log_f.write(str(rel_path) + "\n")

        if i % 50 == 0 or i == len(remaining):
            print(f"  {i}/{len(remaining)} processed")

print(f"\nDone. Detections table: {detections_csv_path}")
print(f"Visuals: {full_vis_dir}")
```

## SECTION 4: Localized Coral Colony Cropping

### 4.1 Cropping every bounding box into patches

Output layout:
  > `Coralnet-Patches` / `<site>` / `<survey>` / `<transect>` / `<image_stem>` / **CoralN.JPG**

Patches numbered by descending confidence within each source image.

```python
PATCHES_ROOT = Path(CORAL_PATCHES_DIR)
PATCH_PADDING_FRAC = 0.0   # e.g. 0.08 for an 8% margin around each box if Stage 2
                            # training wants a bit of context; 0.0 crops exactly
                            # to the box, matching your spec as given.

def get_xyxy(object_prediction):
    """Robust bbox extraction across SAHI versions."""
    bbox = object_prediction.bbox
    if hasattr(bbox, "to_xyxy"):
        return bbox.to_xyxy()
    return bbox.minx, bbox.miny, bbox.maxx, bbox.maxy

patch_log_path = Path(CORAL_PATCHES_DIR, "patch_processed_log.txt")
patch_already_done = set()
if patch_log_path.exists():
    patch_already_done = set(patch_log_path.read_text().splitlines())

patch_manifest_path = Path(CORAL_PATCHES_DIR, "patches_manifest.csv")
write_patch_header = not patch_manifest_path.exists()

no_exif_warnings = []

with open(patch_manifest_path, "a", newline="") as pf, open(patch_log_path, "a") as plog:
    patch_writer = _csv.writer(pf)
    if write_patch_header:
        patch_writer.writerow([
            "source_image_relpath", "patch_relpath", "patch_index",
            "confidence", "x1", "y1", "x2", "y2"
        ])

    remaining = [p for p in cropped_images if str(p.relative_to(OUTPUT_ROOT)) not in patch_already_done]
    print(f"{len(patch_already_done)} source images already patch-processed, {len(remaining)} remaining.")

    for i, img_path in enumerate(remaining, 1):
        rel_path = img_path.relative_to(OUTPUT_ROOT)

        source_img = Image.open(img_path)
        exif_bytes = source_img.info.get("exif")
        if not exif_bytes:
            no_exif_warnings.append(str(rel_path))

        result = sliced_predict_safe(str(img_path), **SAHI_KWARGS)
        predictions = sorted(result.object_prediction_list, key=lambda p: p.score.value, reverse=True)

        if predictions:
            patch_dir = PATCHES_ROOT / rel_path.parent / rel_path.stem
            patch_dir.mkdir(parents=True, exist_ok=True)

            for idx, pred in enumerate(predictions, 1):
                x1, y1, x2, y2 = get_xyxy(pred)
                if PATCH_PADDING_FRAC > 0:
                    w, h = x2 - x1, y2 - y1
                    x1 -= w * PATCH_PADDING_FRAC; x2 += w * PATCH_PADDING_FRAC
                    y1 -= h * PATCH_PADDING_FRAC; y2 += h * PATCH_PADDING_FRAC
                x1, y1 = max(0, x1), max(0, y1)
                x2, y2 = min(source_img.width, x2), min(source_img.height, y2)

                if x2 - x1 < 2 or y2 - y1 < 2:
                    continue  # degenerate box (e.g. clipped at an edge) -- skip, don't crash

                patch = source_img.crop((int(x1), int(y1), int(x2), int(y2)))
                patch_path = patch_dir / f"Coral{idx}.JPG"
                save_kwargs = {"exif": exif_bytes} if exif_bytes else {}
                patch.convert("RGB").save(patch_path, **save_kwargs)

                patch_writer.writerow([
                    str(rel_path), str(patch_path.relative_to(PATCHES_ROOT)), idx,
                    round(pred.score.value, 4), round(x1, 1), round(y1, 1), round(x2, 1), round(y2, 1),
                ])

        plog.write(str(rel_path) + "\n")
        if i % 50 == 0 or i == len(remaining):
            print(f"  {i}/{len(remaining)} source images processed")

print(f"\nDone. Patch manifest: {patch_manifest_path}")
print(f"Patches root: {PATCHES_ROOT}")
if no_exif_warnings:
    print(f"\n[WARNING] {len(no_exif_warnings)} source image(s) had no EXIF to carry forward -- "
          f"crop_coralnet_quadrat.py's save() call likely needs exif=img.info.get('exif') added. "
          f"First few: {no_exif_warnings[:5]}")
```

## SECTION 5: Coral Patches Refinement

### 5.1 Color correction and GrabCut Tightening (combined, re-crop from source)

**Order matters:** color-correct the FULL source image first (gray-world's
assumption only holds at that scale), then tighten each box on the
corrected image (color correction also improves GrabCut's own foreground/
background separation, which depends on color contrast the original cast
was compressing).

```python
APPLY_COLOR_CORRECTION = True
APPLY_GRABCUT_TIGHTENING = True
REFINED_PATCHES_ROOT = Path(REFINED_PATCHES_DIR)  # new, non-destructive
patch_manifest_path = Path(CORAL_PATCHES_DIR, "patches_manifest.csv")

import csv as _csv4

def correct_underwater_color(img_bgr, clip_limit=2.0, target_clip_pct=1.0, hard_ceiling=2.0, hard_floor=0.7):
    """Gray-world white balance + CLAHE on lightness only. Deterministic --
    identical logic applied to every image, nothing randomized.

    GAIN CALCULATION (revised): a first attempt capped every channel's gain
    to a flat [0.8, 1.3] range. That stopped the clipping (good) but tested
    against all 8 curated samples, EVERY one needed >1.3x red correction, so
    every image got flattened to the exact same 1.300 gain regardless of its
    actual site/depth -- silently erasing the per-image adaptation gray-world
    is supposed to provide.

    Fixed properly: each channel's gain is the smaller of (a) the standard
    mean-based gray-world gain, and (b) the largest gain that keeps at most
    ~1% of that channel's own pixels clipping at 255 (derived from that
    channel's own percentile distribution, not a shared constant). Tested
    against all 8 samples spanning 5 stations/3.5-8.0m depth: final gains
    now range 1.19x-1.64x, tracking real depth (Coral Garden S at 4.0m needs
    the least correction, Harka Piloto D at 8.0m needs the most), while
    clipping stays ~1% on every sample -- the per-site variation is real
    again AND the earlier oversaturation is still fixed."""
    b, g, r = cv2.split(img_bgr.astype(np.float64))
    mean_gray = (b.mean() + g.mean() + r.mean()) / 3

    def channel_gain(channel):
        mean_gain = mean_gray / channel.mean()
        if mean_gain > 1.0:
            safe_ceiling = np.percentile(channel, 100 - target_clip_pct)
            clipsafe_gain = 255.0 / safe_ceiling if safe_ceiling > 0 else mean_gain
            return min(mean_gain, clipsafe_gain, hard_ceiling)
        else:
            return max(mean_gain, hard_floor)

    gain_b, gain_g, gain_r = channel_gain(b), channel_gain(g), channel_gain(r)
    b = np.clip(b * gain_b, 0, 255)
    g = np.clip(g * gain_g, 0, 255)
    r = np.clip(r * gain_r, 0, 255)
    balanced = cv2.merge([b, g, r]).astype(np.uint8)
    lab = cv2.cvtColor(balanced, cv2.COLOR_BGR2LAB)
    l, a, b_ch = cv2.split(lab)
    l = cv2.createCLAHE(clipLimit=clip_limit, tileGridSize=(8, 8)).apply(l)
    return cv2.cvtColor(cv2.merge([l, a, b_ch]), cv2.COLOR_LAB2BGR)

def tighten_box_grabcut(image_rgb, x1, y1, x2, y2, expand_frac=0.15, iterations=5, max_roi_dim=800):
    """Downsamples large ROIs before running GrabCut, then scales the result
    back up. Cost scales with pixel count -- without this, the largest (often
    loosest) boxes can take a minute or more each."""
    h, w = image_rgb.shape[:2]
    bw, bh = x2 - x1, y2 - y1
    ex1, ey1 = max(0, int(x1 - bw * expand_frac)), max(0, int(y1 - bh * expand_frac))
    ex2, ey2 = min(w, int(x2 + bw * expand_frac)), min(h, int(y2 + bh * expand_frac))
    roi = image_rgb[ey1:ey2, ex1:ex2]
    roi_h, roi_w = roi.shape[:2]

    scale = min(1.0, max_roi_dim / max(roi_h, roi_w))
    small_roi = cv2.resize(roi, (max(1,int(roi_w*scale)), max(1,int(roi_h*scale)))) if scale < 1.0 else roi

    mask = np.zeros(small_roi.shape[:2], np.uint8)
    bgd_model, fgd_model = np.zeros((1, 65), np.float64), np.zeros((1, 65), np.float64)
    rect = (int((x1 - ex1)*scale), int((y1 - ey1)*scale), int((x2 - x1)*scale), int((y2 - y1)*scale))
    try:
        cv2.grabCut(small_roi, mask, rect, bgd_model, fgd_model, iterations, cv2.GC_INIT_WITH_RECT)
    except cv2.error:
        return x1, y1, x2, y2, False
    fg_mask = np.where((mask == cv2.GC_FGD) | (mask == cv2.GC_PR_FGD), 1, 0).astype(np.uint8)
    ys, xs = np.where(fg_mask > 0)
    if len(xs) == 0 or len(ys) == 0:
        return x1, y1, x2, y2, False
    tx1, tx2 = xs.min()/scale + ex1, xs.max()/scale + ex1
    ty1, ty2 = ys.min()/scale + ey1, ys.max()/scale + ey1
    if (tx2 - tx1) < bw * 0.3 or (ty2 - ty1) < bh * 0.3:
        return x1, y1, x2, y2, False
    return tx1, ty1, tx2, ty2, True

def load_patches_manifest_grouped(manifest_path):
    grouped = defaultdict(list)
    with open(manifest_path, newline="") as f:
        for row in _csv4.DictReader(f):
            grouped[row["source_image_relpath"]].append(row)
    return grouped

def _cell1_selftest():
    checks = []

    try:
        dummy = np.random.randint(0, 255, (50, 50, 3), dtype=np.uint8)
        corrected = correct_underwater_color(dummy)
        assert corrected.shape == dummy.shape and corrected.dtype == np.uint8
        checks.append(("correct_underwater_color runs correctly", True, ""))
    except Exception as e:
        checks.append(("correct_underwater_color runs correctly", False, f"{type(e).__name__}: {e}"))

    try:
        dummy_rgb = np.random.randint(0, 255, (100, 100, 3), dtype=np.uint8)
        result = tighten_box_grabcut(dummy_rgb, 20, 20, 80, 80)
        assert len(result) == 5
        checks.append(("tighten_box_grabcut runs correctly", True, ""))
    except Exception as e:
        checks.append(("tighten_box_grabcut runs correctly", False, f"{type(e).__name__}: {e}"))

    manifest_ok = patch_manifest_path.exists()
    checks.append((f"patches_manifest.csv found at {patch_manifest_path}", manifest_ok,
                    "" if manifest_ok else "Preview/Full Batch will fail without this"))

    output_root_ok = OUTPUT_ROOT.exists()
    checks.append((f"OUTPUT_ROOT found at {OUTPUT_ROOT}", output_root_ok,
                    "" if output_root_ok else "Preview/Full Batch will fail reading source images"))

    print("=" * 60)
    print("CELL 1 SELF-CHECK")
    print("=" * 60)
    all_passed = True
    for label, passed, detail in checks:
        status = "OK  " if passed else "FAIL"
        print(f"[{status}] {label}" + (f" -- {detail}" if detail else ""))
        all_passed = all_passed and passed
    print("-" * 60)
    print(f"APPLY_COLOR_CORRECTION   = {APPLY_COLOR_CORRECTION}")
    print(f"APPLY_GRABCUT_TIGHTENING = {APPLY_GRABCUT_TIGHTENING}")
    print(f"REFINED_PATCHES_ROOT     = {REFINED_PATCHES_ROOT}")
    print("=" * 60)
    print("Cell 1 ready -- proceed to Preview." if all_passed
          else "Unresolved issues above -- fix before running Preview or Full Batch.")

_cell1_selftest()
```

### 5.2 PREVIEW (2 source images, before committing to all 894)

```python
N_PREVIEW_PATCHES = 10

grouped = load_patches_manifest_grouped(patch_manifest_path)
all_patch_rows = [(src, row) for src, rows in grouped.items() for row in rows]
preview_rows = random.sample(all_patch_rows, min(N_PREVIEW_PATCHES, len(all_patch_rows)))

_source_cache = {}
def _get_working_image(src_rel):
    if src_rel not in _source_cache:
        source_bgr = cv2.imread(str(OUTPUT_ROOT / src_rel))
        if source_bgr is None:
            _source_cache[src_rel] = None
        else:
            original_rgb = cv2.cvtColor(source_bgr, cv2.COLOR_BGR2RGB)
            working_bgr = correct_underwater_color(source_bgr) if APPLY_COLOR_CORRECTION else source_bgr
            _source_cache[src_rel] = (original_rgb, cv2.cvtColor(working_bgr, cv2.COLOR_BGR2RGB))
    return _source_cache[src_rel]

fig, axes = plt.subplots(len(preview_rows), 2, figsize=(8, 4 * len(preview_rows)))
row_i = 0
for src_rel, row in preview_rows:
    imgs = _get_working_image(src_rel)
    if imgs is None:
        print(f"[skip] could not read: {(OUTPUT_ROOT / src_rel).resolve()}")
        continue
    original_rgb, working_rgb = imgs
    x1, y1, x2, y2 = float(row["x1"]), float(row["y1"]), float(row["x2"]), float(row["y2"])
    if APPLY_GRABCUT_TIGHTENING:
        tx1, ty1, tx2, ty2, _ = tighten_box_grabcut(working_rgb, x1, y1, x2, y2)
    else:
        tx1, ty1, tx2, ty2 = x1, y1, x2, y2

    axes[row_i][0].imshow(original_rgb[int(y1):int(y2), int(x1):int(x2)])
    axes[row_i][0].set_title(f"{Path(src_rel).stem} #{row['patch_index']} -- original"); axes[row_i][0].axis("off")
    axes[row_i][1].imshow(working_rgb[int(ty1):int(ty2), int(tx1):int(tx2)])
    axes[row_i][1].set_title("corrected + tightened"); axes[row_i][1].axis("off")
    row_i += 1

plt.tight_layout()
plt.show()
```

### 5.3 Full Batch Recropping

```python
refine_log_path = Path(REFINED_PATCHES_DIR, "refine_processed_log.txt")
refine_already_done = set(refine_log_path.read_text().splitlines()) if refine_log_path.exists() else set()

refine_manifest_path = Path(REFINED_PATCHES_DIR, "patches_manifest_refined.csv")
write_refine_header = not refine_manifest_path.exists()

grouped = load_patches_manifest_grouped(patch_manifest_path)
remaining_sources = [s for s in grouped if s not in refine_already_done]
print(f"{len(refine_already_done)} source images already refined, {len(remaining_sources)} remaining.")

grabcut_success_count, grabcut_fallback_count = 0, 0

with open(refine_manifest_path, "a", newline="") as rf, open(refine_log_path, "a") as rlog:
    refine_writer = _csv4.writer(rf)
    if write_refine_header:
        refine_writer.writerow([
            "source_image_relpath", "patch_relpath", "patch_index", "confidence",
            "orig_x1", "orig_y1", "orig_x2", "orig_y2",
            "refined_x1", "refined_y1", "refined_x2", "refined_y2", "grabcut_applied",
        ])

    for i, src_rel in enumerate(remaining_sources, 1):
        source_path = OUTPUT_ROOT / src_rel
        exif_bytes = Image.open(source_path).info.get("exif")

        source_bgr = cv2.imread(str(source_path))
        if source_bgr is None:
            print(f"[skip] could not read {src_rel}")
            rlog.write(src_rel + "\n")
            continue

        working_bgr = correct_underwater_color(source_bgr) if APPLY_COLOR_CORRECTION else source_bgr
        working_rgb = cv2.cvtColor(working_bgr, cv2.COLOR_BGR2RGB)
        h, w = working_rgb.shape[:2]

        for row in grouped[src_rel]:
            x1, y1, x2, y2 = float(row["x1"]), float(row["y1"]), float(row["x2"]), float(row["y2"])
            if APPLY_GRABCUT_TIGHTENING:
                tx1, ty1, tx2, ty2, applied = tighten_box_grabcut(working_rgb, x1, y1, x2, y2)
            else:
                tx1, ty1, tx2, ty2, applied = x1, y1, x2, y2, False

            tx1, ty1 = max(0, tx1), max(0, ty1)
            tx2, ty2 = min(w, tx2), min(h, ty2)
            if tx2 - tx1 < 2 or ty2 - ty1 < 2:
                continue

            grabcut_success_count += applied
            grabcut_fallback_count += not applied

            patch_path = REFINED_PATCHES_ROOT / row["patch_relpath"]
            patch_path.parent.mkdir(parents=True, exist_ok=True)
            patch_pil = Image.fromarray(working_rgb[int(ty1):int(ty2), int(tx1):int(tx2)])
            save_kwargs = {"exif": exif_bytes} if exif_bytes else {}
            patch_pil.convert("RGB").save(patch_path, **save_kwargs)

            refine_writer.writerow([
                src_rel, row["patch_relpath"], row["patch_index"], row["confidence"],
                round(x1,1), round(y1,1), round(x2,1), round(y2,1),
                round(tx1,1), round(ty1,1), round(tx2,1), round(ty2,1), applied,
            ])

        rlog.write(src_rel + "\n")
        if i % 50 == 0 or i == len(remaining_sources):
            print(f"  {i}/{len(remaining_sources)} source images refined")

print(f"\nDone. Refined manifest: {refine_manifest_path}")
print(f"Refined patches root: {REFINED_PATCHES_ROOT}")
if APPLY_GRABCUT_TIGHTENING:
    total = grabcut_success_count + grabcut_fallback_count
    print(f"GrabCut tightened {grabcut_success_count}/{total} patches "
          f"({grabcut_fallback_count} fell back to the original box).")
```

### 5.4 Refined Patches Audit

After the full-batch refinement, coral patches were manually reviewed and any
`Coral[N].JPG` that didn't actually show a coral colony (sand, rubble,
unidentifiable substrate, etc.) were deleted directly from `Refined-Patches`.

This section doesn't touch any files — it's read-only. It compares
`patches_manifest_refined.csv` (everything the refinement run *produced*)
against what's still physically present on disk (everything that *survived*
manual curation), and reports the difference.

```python
import re

def parse_directory_folder_number(patch_relpath):
    """Splits a patch_relpath like '.../GOPR3088/Coral3.JPG' into its
    directory, the GOPR-named source folder, and the patch number."""
    p = Path(patch_relpath)
    directory = str(p.parent)
    gopr_folder = p.parent.name
    m = re.search(r"Coral(\d+)", p.stem, re.IGNORECASE)
    patch_number = int(m.group(1)) if m else None
    return directory, gopr_folder, patch_number

refine_manifest_path = Path(REFINED_PATCHES_DIR, "patches_manifest_refined.csv")

with open(refine_manifest_path, newline="") as f:
    manifest_rows = list(csv.DictReader(f))

deleted_rows = [
    row for row in manifest_rows
    if not (REFINED_PATCHES_ROOT / row["patch_relpath"]).exists()
]

print("=" * 70)
print("5.2 PATCH CURATION AUDIT")
print("=" * 70)
print(f"Original refined patches (per patches_manifest_refined.csv): {len(manifest_rows)}")
print(f"Deleted during manual curation:                              {len(deleted_rows)}")
print(f"Remaining on disk:                                           {len(manifest_rows) - len(deleted_rows)}")
print()

# ---- 1. Detailed listing: directory + specific patch number(s) deleted ----
by_directory = defaultdict(list)
for row in deleted_rows:
    directory, gopr_folder, patch_number = parse_directory_folder_number(row["patch_relpath"])
    by_directory[directory].append(patch_number)

print("-" * 70)
print("DELETED PATCHES BY DIRECTORY")
print("-" * 70)
if not by_directory:
    print("(none -- every patch in the manifest is still present on disk)")
for directory in sorted(by_directory):
    numbers = sorted(n for n in by_directory[directory] if n is not None)
    print(f"{directory}")
    print(f"    Deleted: {', '.join(f'Coral{n}' for n in numbers)}")

# ---- 2. Count of deletions per GOPR (source-image) folder ----
by_gopr_folder = defaultdict(int)
for row in deleted_rows:
    _, gopr_folder, _ = parse_directory_folder_number(row["patch_relpath"])
    by_gopr_folder[gopr_folder] += 1

print()
print("-" * 70)
print("DELETED COUNT PER GOPR FOLDER")
print("-" * 70)
for folder in sorted(by_gopr_folder):
    print(f"  {folder:20s} {by_gopr_folder[folder]} deleted")

# ---- 3. Grand totals, for discrepancy tracking ----
print()
print("-" * 70)
print("SUMMARY (for discrepancy purposes)")
print("-" * 70)
print(f"Original patch count (patches_manifest_refined.csv): {len(manifest_rows)}")
print(f"Deleted during manual curation:                      {len(deleted_rows)}")
print(f"Remaining on disk:                                   {len(manifest_rows) - len(deleted_rows)}")

# ---- 4. Downloadable CSV of everything deleted ----
audit_csv_path = REFINED_PATCHES_ROOT / "deleted_patches_audit.csv"
with open(audit_csv_path, "w", newline="") as f:
    writer = csv.writer(f)
    writer.writerow([
        "directory", "gopr_folder", "patch_number", "patch_relpath",
        "source_image_relpath", "confidence", "grabcut_applied",
    ])
    for row in deleted_rows:
        directory, gopr_folder, patch_number = parse_directory_folder_number(row["patch_relpath"])
        writer.writerow([
            directory, gopr_folder, patch_number, row["patch_relpath"],
            row["source_image_relpath"], row.get("confidence", ""),
            row.get("grabcut_applied", ""),
        ])

print(f"\nFull audit CSV written to: {audit_csv_path}")
```
