## Section 10: Fine Tuning

### 10.1 Preflight check for coral-quadrat-2-final + Test_Run_V6-fixed weights

> (*One Time Only. No need to run anymore.*)
> 
**Purpose:**

Verifies the cleaned coral-quadrat-2 dataset and the source checkpoint are both sound BEFORE spending GPU time on step2_finetune.py. Catches the two failure modes that have already cost time in this project:
1. Corrupted / mismatched class names -- the XD38M lesson: always verify model.names directly, never assume data.yaml is correct.
2. Malformed or out-of-range dataset entries that would otherwise silently degrade training instead of throwing a clear error.

Run this and read the full report before proceeding to ``step2_finetune.py``.

```python
DATASET_ROOT = Path(FNL_CORAL_QUAD_DIR)
BEST_WEIGHTS_PATH = Path(BEST_WEIGHTS_PATH)
SPLITS = ["train", "valid", "test"]
EXPECTED_CLASS_NAME = "coral"
EXPECTED_NC = 1
OVERSIZED_WH_THRESHOLD = 0.85  # re-check that step0's cleanup actually took effect
# ===================================================================================

issues = []
warnings = []


def check_data_yaml():
    yaml_path = DATASET_ROOT / "data.yaml"
    print(f"\n--- Checking {yaml_path} ---")
    if not yaml_path.exists():
        issues.append(f"data.yaml not found at {yaml_path}")
        return None
    with open(yaml_path) as f:
        cfg = yaml.safe_load(f)
    nc = cfg.get("nc")
    names = cfg.get("names")
    print(f"nc: {nc}")
    print(f"names: {names}")
    if nc != EXPECTED_NC:
        issues.append(f"Expected nc={EXPECTED_NC}, found nc={nc}")
    names_list = list(names.values()) if isinstance(names, dict) else names
    if names_list != [EXPECTED_CLASS_NAME]:
        issues.append(f"Expected names=['{EXPECTED_CLASS_NAME}'], found names={names_list}")
    return cfg


def check_model_weights():
    print(f"\n--- Checking model weights: {BEST_WEIGHTS_PATH} ---")
    if not BEST_WEIGHTS_PATH.exists():
        issues.append(f"Weights file not found: {BEST_WEIGHTS_PATH}")
        return
    try:
        from ultralytics import YOLO
        model = YOLO(str(BEST_WEIGHTS_PATH))
        print(f"model.names: {model.names}")
        names_list = list(model.names.values())
        if names_list != [EXPECTED_CLASS_NAME]:
            issues.append(
                f"Loaded model reports names={names_list}, expected ['{EXPECTED_CLASS_NAME}']. "
                f"This is exactly the XD38M corruption pattern seen earlier in the project -- "
                f"do not proceed until this is resolved."
            )
        else:
            print("Class names confirmed clean.")
    except Exception as e:
        issues.append(f"Failed to load model weights: {e}")


def check_split(split: str):
    print(f"\n--- Checking split: {split} ---")
    img_dir = DATASET_ROOT / split / "images"
    lbl_dir = DATASET_ROOT / split / "labels"

    if not img_dir.exists() or not lbl_dir.exists():
        issues.append(f"Missing images/ or labels/ folder for split '{split}'")
        return

    img_stems = {p.stem for p in img_dir.glob("*") if p.suffix.lower() in (".jpg", ".jpeg", ".png")}
    lbl_stems = {p.stem for p in lbl_dir.glob("*.txt")}

    missing_labels = img_stems - lbl_stems
    orphan_labels = lbl_stems - img_stems

    if missing_labels:
        warnings.append(f"[{split}] {len(missing_labels)} images have no matching label file "
                         f"(e.g. {sorted(missing_labels)[:3]})")
    if orphan_labels:
        warnings.append(f"[{split}] {len(orphan_labels)} label files have no matching image "
                         f"(e.g. {sorted(orphan_labels)[:3]})")

    total_boxes = 0
    oversized_remaining = 0
    out_of_range = 0
    wrong_class = 0
    corrupt_images = 0
    box_widths, box_heights = [], []

    for stem in sorted(img_stems & lbl_stems):
        lbl_file = lbl_dir / f"{stem}.txt"
        for line in lbl_file.read_text().splitlines():
            if not line.strip():
                continue
            parts = line.split()
            if len(parts) < 5:
                warnings.append(f"[{split}] malformed label line in {stem}.txt")
                continue
            cls = int(parts[0])
            cx, cy, bw, bh = map(float, parts[1:5])
            total_boxes += 1
            box_widths.append(bw)
            box_heights.append(bh)

            if cls != 0:
                wrong_class += 1
            if not (0.0 <= cx <= 1.0 and 0.0 <= cy <= 1.0 and 0.0 < bw <= 1.0 and 0.0 < bh <= 1.0):
                out_of_range += 1
            if bw > OVERSIZED_WH_THRESHOLD and bh > OVERSIZED_WH_THRESHOLD:
                oversized_remaining += 1

    # Sample a handful of images to confirm they actually open correctly
    sample_stems = sorted(img_stems & lbl_stems)[:25]
    for stem in sample_stems:
        candidates = list(img_dir.glob(f"{stem}.*"))
        if not candidates:
            continue
        try:
            with Image.open(candidates[0]) as im:
                im.verify()
        except Exception:
            corrupt_images += 1
            warnings.append(f"[{split}] could not open image for '{stem}' -- possibly corrupt")

    print(f"Images: {len(img_stems)} | Labels: {len(lbl_stems)} | Total boxes: {total_boxes}")
    if box_widths:
        print(f"Avg box size (normalized): w={sum(box_widths)/len(box_widths):.3f}, "
              f"h={sum(box_heights)/len(box_heights):.3f}")
    print(f"Boxes with wrong class (!=0): {wrong_class}")
    print(f"Boxes with out-of-range coordinates: {out_of_range}")
    print(f"Remaining oversized (habitat-scale) boxes: {oversized_remaining}")
    print(f"Sampled images that failed to open: {corrupt_images} / {len(sample_stems)}")

    if wrong_class:
        issues.append(f"[{split}] {wrong_class} boxes have class != 0 -- remap did not fully apply")
    if out_of_range:
        issues.append(f"[{split}] {out_of_range} boxes have out-of-range normalized coordinates")
    if oversized_remaining:
        issues.append(f"[{split}] {oversized_remaining} oversized boxes still present -- "
                       f"did step0_cleanup_oversized_boxes.py run on THIS folder?")


def main():
    print("=" * 70)
    print("PREFLIGHT CHECK -- coral-quadrat-2-cleaned + Test_Run_V6-fixed weights")
    print("=" * 70)

    check_data_yaml()
    check_model_weights()
    for split in SPLITS:
        check_split(split)

    print("\n" + "=" * 70)
    print("SUMMARY")
    print("=" * 70)
    if warnings:
        print(f"\n{len(warnings)} WARNING(S):")
        for w in warnings:
            print(f"  - {w}")
    if issues:
        print(f"\n{len(issues)} BLOCKING ISSUE(S) -- resolve before fine-tuning:")
        for i in issues:
            print(f"  - {i}")
        print("\nDo NOT proceed to step2_finetune.py until these are resolved.")
    else:
        print("\nNo blocking issues found. Safe to proceed to step2_finetune.py.")


if __name__ == "__main__":
    main()
```

### 10.2 Calibration Experiment

> (grid version): Systematic ablation over freeze depth / LR / augmentation

**Purpose:**
    Runs ONE experiment per execution, selected via ACTIVE_EXPERIMENT below. All
    experiments always start fresh from Test_Run_V6-fixed's best.pt (Exp5 baseline)
    -- never chain off a previous fine-tuned checkpoint, or you can't cleanly
    attribute results to the hyperparameter change being tested.
 
**Workflow:**
1. Set `ACTIVE_EXPERIMENT = "Exp8"`, run this script.
2. Set `ACTIVE_EXPERIMENT = "Exp9"`, run again. ... and so on through Tier 1.
3. Once Tier 1 results are in, update `BEST_FREEZE_FROM_TIER1` below to whichever
       freeze depth gave the best combined result (see the diagnosis criteria
       from the last round: XD38M retention + CoralNet confidence-sweep quality
       + coral-quadrat-2 gain -- not just one metric in isolation).
4. Run Exp13, Exp14 (Tier 2) -- they automatically use BEST_FREEZE_FROM_TIER1.
5. Update `BEST_LR_FROM_TIER2`, run Exp15, Exp16 (Tier 3).
6. Update `BEST_AUG_FROM_TIER3`, run Exp17, Exp18 (Tier 4).
7. Run Exp19 (fixed values, no placeholders needed) whenever convenient.
 
Each run saves checkpoints every `SAVE_PERIOD` epochs (not just best/last), so
    you can retroactively inspect intermediate epochs post-hoc if the
    EarlyStopping-selected best.pt (chosen using ONLY coral-quadrat-2 fitness)
    turns out not to be the best combined checkpoint.



```python
"""
Calibration.py -- reconfigured after Tier 1 (Test_Run_V1..V5) diagnosis
=========================================================================
WHAT CHANGED FROM YOUR VERSION, AND WHY
-----------------------------------------------------------------------
1. BEST_FREEZE_FROM_TIER1 = 0 is now CONFIRMED, not assumed. The Tier 1
   evaluation data (all_experiments_validation_comparison.csv) shows
   Exp5_finetuned (freeze=0) beats every partial-freeze depth on
   coral-quadrat-2 val -- your actual domain-adaptation target -- on
   mAP50 (0.611 vs 0.505 for the next best), precision (0.636), and
   recall (0.548) simultaneously. See the diagnosis for the full
   trade-off curve against XD38M retention.
 
2. BEST_LR_FROM_TIER2 and BEST_AUG_FROM_TIER3 are marked PROVISIONAL.
   Nothing in the data you've shared so far (only Exp1-5 / Tier 1 have
   results) actually validates 0.00002 or "reduced" as winners --
   those presets exist in CONFIG_PRESETS ready to run, but haven't been
   run yet. Treat the values below as placeholders you are ABOUT to
   test, not conclusions you can cite in your methodology chapter yet.
 
3. ACTIVE_EXPERIMENT now defaults to "modelCalib_Exp6" -- the first
   real Tier 2 run (LR=0.00005, freeze=0 fixed via the now-confirmed
   BEST_FREEZE_FROM_TIER1). Re-running Exp1-5 would just repeat Tier 1.
 
4. A lightweight tier-guard was added (see TIER_VALIDATED below) that
   prints a warning -- not a hard stop -- if you try to launch a Tier 3+
   preset before Tier 2 has actually produced results. This is meant to
   catch the exact trap described above, not to slow you down.
 
5. This file reconstructs the training-launch call (imports, run
   naming, model.train(...)) that wasn't included in the documentation
   log excerpt -- only the config dictionaries were. The config section
   below is otherwise unchanged from your version. Diff this against
   your actual local Calibration.py before overwriting anything, and
   double check the optimizer line against what your XD38M baseline
   training actually used.
"""
 
 
# ============================== FIXED CONFIG -- do not vary these ==================
BASE_WEIGHTS = Path(BEST_WEIGHTS_PATH)
FINETUNE_YAML_PATH = Path(CORAL_QUAD_YAML_PATH)
FINETUNE_RUNS_DIR = Path(RUNS_DIR)
EPOCHS = 80
PATIENCE = 20
SAVE_PERIOD = 5   # keep an intermediate checkpoint every 5 epochs for post-hoc review
# =====================================================================================
 
# ============================== TIER VALIDATION STATUS -- new ======================
# Tracks what's actually been empirically confirmed by evaluated results, vs. what's
# still a placeholder waiting on a run. Update these by hand as each tier finishes
# and gets evaluated (Evaluation.py's EXPERIMENTS dict uncommented + re-run).
TIER_VALIDATED = {
    1: True,   # freeze depth -- CONFIRMED by Test_Run_V1..V5 (Exp1-5) results
    2: True,  # learning rate -- NOT yet run/evaluated (Exp6, Exp7, Exp8)
    3: True,  # augmentation  -- depends on Tier 2's winner, not yet run
    4: False,  # warmup/batch  -- depends on Tier 3's winner, not yet run
    5: False,  # fine LR micro-sweep -- depends on Tier 2
    6: False,  # zero-augmentation extreme -- depends on Tier 2/3
}
 
EXPERIMENT_TIER = {
    "modelCalib_Exp1": 1, "modelCalib_Exp2": 1, "modelCalib_Exp3": 1,
    "modelCalib_Exp4": 1, "modelCalib_Exp5": 1,
    "modelCalib_Exp6": 2, "modelCalib_Exp7": 2, "modelCalib_Exp8": 2,
    "modelCalib_Exp9": 3, "modelCalib_Exp10": 3,
    "modelCalib_Exp11": 4, "modelCalib_Exp12": 4,
    "modelCalib_Exp13": 5, "modelCalib_Exp14": 5,
    "modelCalib_Exp15": 6,
}
 
 
def check_tier_readiness(experiment_label: str) -> None:
    """Warn (don't block) if launching a preset whose upstream tier is unconfirmed."""
    tier = EXPERIMENT_TIER.get(experiment_label)
    if tier is None:
        return
    prior_tiers = [t for t in TIER_VALIDATED if t < tier]
    unconfirmed = [t for t in prior_tiers if not TIER_VALIDATED[t]]
    if unconfirmed:
        print(f"[WARNING] {experiment_label} is Tier {tier}, but Tier(s) "
              f"{unconfirmed} have not been confirmed by evaluated results yet. "
              f"Any 'BEST_*_FROM_TIER{unconfirmed[0]}' value it depends on is "
              f"still a placeholder. Consider running the missing tier first.")
 
 
# ============================== SETTLED VALUES ======================================
BEST_FREEZE_FROM_TIER1 = 0        # CONFIRMED: full unfreeze beat every partial-freeze
                                   # depth tested on coral-quadrat-2 val (Tier 1 data)
BEST_LR_FROM_TIER2 = 0.0001      # PROVISIONAL -- placeholder pending Tier 2 results
BEST_AUG_FROM_TIER3 = "moderate"   # PROVISIONAL -- placeholder pending Tier 3 results
# =====================================================================================
 
# ============================== PICK ONE EXPERIMENT PER RUN =========================
ACTIVE_EXPERIMENT = "modelCalib_Exp11"   # next real step: start of the Tier 2 LR sweep
# =====================================================================================
 
AUGMENTATION_PROFILES = {
    "reduced": dict(
        mosaic=0.0, mixup=0.0, cutmix=0.0, erasing=0.1, auto_augment=None,
        hsv_h=0.01, hsv_s=0.5, hsv_v=0.3, translate=0.05, scale=0.3,
        fliplr=0.5, flipud=0.5,
    ),
    "moderate": dict(
        mosaic=0.5, mixup=0.05, cutmix=0.0, erasing=0.15, auto_augment=None,
        hsv_h=0.01, hsv_s=0.6, hsv_v=0.35, translate=0.08, scale=0.4,
        fliplr=0.5, flipud=0.5,
    ),
    "full": dict(
        mosaic=1.0, mixup=0.1, cutmix=0.0, erasing=0.2, auto_augment="randaugment",
        hsv_h=0.015, hsv_s=0.7, hsv_v=0.4, translate=0.1, scale=0.5,
        fliplr=0.5, flipud=0.5,
    ),
    "none": dict(
        mosaic=0.0, mixup=0.0, cutmix=0.0, erasing=0.0, auto_augment=None,
        hsv_h=0.0, hsv_s=0.0, hsv_v=0.0, translate=0.0, scale=0.0,
        fliplr=0.5, flipud=0.5,
    ),
}
 
CONFIG_PRESETS = {
    # ---------------- TIER 1: freeze-depth sweep -- COMPLETE, confirmed -------------
    "modelCalib_Exp1": dict(freeze=13, lr0=0.0001, aug="reduced", batch=8, warmup_epochs=3),
    "modelCalib_Exp2": dict(freeze=16, lr0=0.0001, aug="reduced", batch=8, warmup_epochs=3),
    "modelCalib_Exp3": dict(freeze=19, lr0=0.0001, aug="reduced", batch=8, warmup_epochs=3),
    "modelCalib_Exp4": dict(freeze=22, lr0=0.0001, aug="reduced", batch=8, warmup_epochs=3),
    "modelCalib_Exp5": dict(freeze=0,  lr0=0.0001, aug="reduced", batch=8, warmup_epochs=3),
 
    # ---------------- TIER 2: learning-rate sweep -- RUN THIS NEXT ------------------
    "modelCalib_Exp6": dict(freeze=BEST_FREEZE_FROM_TIER1, lr0=0.00005, aug="reduced", batch=8, warmup_epochs=3),
    "modelCalib_Exp7": dict(freeze=BEST_FREEZE_FROM_TIER1, lr0=0.00002, aug="reduced", batch=8, warmup_epochs=3),
    "modelCalib_Exp8": dict(freeze=BEST_FREEZE_FROM_TIER1, lr0=0.0002,  aug="reduced", batch=8, warmup_epochs=3),
 
    # ---------------- TIER 3: augmentation sweep -------------------------------------
    "modelCalib_Exp9":  dict(freeze=BEST_FREEZE_FROM_TIER1, lr0=BEST_LR_FROM_TIER2, aug="full",     batch=8, warmup_epochs=3),
    "modelCalib_Exp10": dict(freeze=BEST_FREEZE_FROM_TIER1, lr0=BEST_LR_FROM_TIER2, aug="moderate", batch=8, warmup_epochs=3),
 
    # ---------------- TIER 4: secondary levers -- optional, lower priority ----------
    "modelCalib_Exp11": dict(freeze=BEST_FREEZE_FROM_TIER1, lr0=BEST_LR_FROM_TIER2, aug=BEST_AUG_FROM_TIER3, batch=8,  warmup_epochs=10),
    "modelCalib_Exp12": dict(freeze=BEST_FREEZE_FROM_TIER1, lr0=BEST_LR_FROM_TIER2, aug=BEST_AUG_FROM_TIER3, batch=16, warmup_epochs=3),
 
    # ---------------- TIER 5: fine LR micro-sweep around Exp7's candidate 2e-5 ------
    "modelCalib_Exp13": dict(freeze=0, lr0=0.00001, aug="reduced", batch=8, warmup_epochs=3),
    "modelCalib_Exp14": dict(freeze=0, lr0=0.00003, aug="reduced", batch=8, warmup_epochs=3),
 
    # ---------------- TIER 6: zero-augmentation extreme -- optional ------------------
    "modelCalib_Exp15": dict(freeze=0, lr0=BEST_LR_FROM_TIER2, aug="none", batch=8, warmup_epochs=3),
 
    # ---------------- OPTIONAL: freeze x low-LR interaction check --------------------
    # Not in the original grid. Tier 1 used a single lr0=0.0001 for every freeze depth,
    # so it never asked whether a *partial* freeze at Tier 2's low LR could match
    # freeze=0's coral-quadrat-2 score while keeping more XD38M capability. Cheap to
    # run (~5 min) given your hardware -- worth it before writing the freeze depth
    # conclusion into your thesis as final. Run only after Tier 2 is confirmed and
    # BEST_LR_FROM_TIER2 is a real value, not the current placeholder.
    "modelCalib_Exp1_lowLR_check": dict(freeze=13, lr0=0.00002, aug="reduced", batch=8, warmup_epochs=3),
}
# =====================================================================================
 
 
def derive_run_name(experiment_label: str) -> str:
    """Test_Run_V{N} where N is the integer suffix of the experiment label.
    Matches the existing 'Exp1' -> 'Test_Run_V1' convention. Non-numeric
    suffixes (e.g. the interaction-check preset) fall back to the label itself."""
    import re
    match = re.search(r"(\d+)$", experiment_label)
    if match:
        return f"Test_Run_V{match.group(1)}"
    return experiment_label
 
 
def main():
    check_tier_readiness(ACTIVE_EXPERIMENT)
    
    if ACTIVE_EXPERIMENT not in CONFIG_PRESETS:
        raise ValueError(f"Unknown experiment '{ACTIVE_EXPERIMENT}'. "
                          f"Choose from: {list(CONFIG_PRESETS.keys())}")
 
    cfg = CONFIG_PRESETS[ACTIVE_EXPERIMENT]
    exp_number = "".join(ch for ch in ACTIVE_EXPERIMENT if ch.isdigit())
    run_name = derive_run_name(ACTIVE_EXPERIMENT)
 
    print(f"Loading base weights from: {BASE_WEIGHTS}  (always the untouched baseline checkpoint -- never chained)")
    model = YOLO(str(BASE_WEIGHTS))
 
    print(f"model.names: {model.names}")
    if list(model.names.values()) != ["coral"]:
        raise RuntimeError(
            f"Loaded model does not report a clean single 'coral' class "
            f"(got {model.names}). Stopping before fine-tuning -- investigate first."
        )
 
    aug_kwargs = AUGMENTATION_PROFILES[cfg["aug"]]
 
    print(f"\n{'='*60}")
    print(f"RUNNING {ACTIVE_EXPERIMENT} -> {run_name}")
    print(f"  freeze={cfg['freeze']}  lr0={cfg['lr0']}  aug={cfg['aug']}  "
          f"batch={cfg['batch']}  warmup_epochs={cfg['warmup_epochs']}")
    print(f"{'='*60}\n")
 
    model.train(
        data=str(FINETUNE_YAML_PATH),
        epochs=EPOCHS,
        imgsz=640,
        batch=cfg["batch"],
        patience=PATIENCE,
        device=0,
        optimizer="AdamW",
        lr0=cfg["lr0"],
        freeze=cfg["freeze"],
        warmup_epochs=cfg["warmup_epochs"],
        save_period=SAVE_PERIOD,
        project=str(FINETUNE_RUNS_DIR),
        name=run_name,
        verbose=True,
        plots=True,
        **aug_kwargs,
    )
 
    print(f"\n{ACTIVE_EXPERIMENT} complete!")
    print(f"Best weights: {FINETUNE_RUNS_DIR / run_name / 'weights' / 'best.pt'}")
    print(f"Intermediate checkpoints (every {SAVE_PERIOD} epochs) also saved for post-hoc review.")
    print(f"\nAdd this checkpoint to evaluation.py's EXPERIMENTS dict, then move on to the next preset.")
 
 
if __name__ == "__main__":
    main()
```

### 10.3 Evaluation

> Batch-evaluate ALL grid-search checkpoints in one pass

**Purpose:**
    Extends `step3_evaluate.py's` logic from a fixed Exp5-vs-Exp6 pair to an
    arbitrary number of checkpoints. Add every experiment you've finished
    training to the EXPERIMENTS dict below and run this once -- it produces
    one consolidated CSV for validation-split comparison and one for the
    CoralNet confidence sweep, plus a rough ranked summary to help you spot
    the most promising configs at a glance.
 
    Missing checkpoints (experiments you haven't run yet) are skipped with a
    warning rather than crashing the whole batch.


```python
"""
EVALUATION.PY -- compares checkpoints on CQ2 val, XD38M val, an optional secondary
labeled dataset, and a CoralNet SAHI confidence sweep -- then writes ONE clean report.
=====================================================================================
Uses the directory setup already run in your notebook (DATA_DIR, RUNS_DIR,
CORAL_QUAD_YAML_PATH, YAML_PATH, CORALNET_TEST_DIR, BEST_WEIGHTS_PATH). Run your
directory-setup cell first, then this one.
 
WHAT CHANGED THIS VERSION, AND WHY
-----------------------------------
1. SAHI merge-quality fix (addresses the Exp9/Exp10 duplicate-box clutter):
   SAHI silently downgrades its merge algorithm from GREEDYNMM/IOS (good at
   catching nested/duplicate boxes) to plain NMS/IOU whenever the model's
   confidence threshold is set very low -- which is exactly what the confidence
   sweep needs to do to capture the full distribution. Rather than rely on an
   unverified override parameter, this version runs TWO separate SAHI passes:
     - a low-threshold (0.01) pass, used ONLY for the confidence-band
       statistics table -- if SAHI downgrades its merge quality here, it
       doesn't matter, since this pass never feeds the visuals or any
       "real detection count".
     - a pass at VISUAL_DISPLAY_THRESHOLD (comfortably above SAHI's low-
       confidence cutoff), which is what actually gets exported as visuals and
       counted as "real" detections -- at this threshold GREEDYNMM/IOS is
       honored properly, with no workaround needed.
   A best-effort attempt at passing force_postprocess_type=True is also made
   (some SAHI versions may support it on this call path), but the script does
   NOT depend on it -- it degrades gracefully if unsupported.
2. No more misleading "combined" heuristic score. Real per-split metrics are
   reported plainly; a secondary labeled dataset (if configured) is the
   strongest available signal, not a blended number.
3. Report generation is now built into this same script -- no separate file
   to run. One file in, one clean report out.
 
Requirements: pip install ultralytics sahi pandas --quiet
"""
 
# ============================== CONFIG -- EDIT THESE ==============================
EXPERIMENTS = {
    "Exp5_baseline":   Path(BEST_WEIGHTS_PATH),
    # "Exp1_finetuned":  Path(RUNS_DIR, "Test_Run_V1", "weights", "best.pt"),
    # "Exp2_finetuned":  Path(RUNS_DIR, "Test_Run_V2", "weights", "best.pt"),
    # "Exp3_finetuned":  Path(RUNS_DIR, "Test_Run_V3", "weights", "best.pt"),
    # "Exp4_finetuned":  Path(RUNS_DIR, "Test_Run_V4", "weights", "best.pt"),
    "Exp5_finetuned":  Path(RUNS_DIR, "Test_Run_V5", "weights", "best.pt"),
    # "Exp6_finetuned":  Path(RUNS_DIR, "Test_Run_V6", "weights", "best.pt"),
    "Exp7_finetuned":  Path(RUNS_DIR, "Test_Run_V7", "weights", "best.pt"),
    # "Exp8_finetuned":  Path(RUNS_DIR, "Test_Run_V8", "weights", "best.pt"),
    "Exp9_finetuned":  Path(RUNS_DIR, "Test_Run_V9", "weights", "best.pt"),
    "Exp10_finetuned": Path(RUNS_DIR, "Test_Run_V10", "weights", "best.pt"),
    # "Exp11_finetuned": Path(RUNS_DIR, "Test_Run_V11", "weights", "best.pt"),
    # "Exp12_finetuned": Path(RUNS_DIR, "Test_Run_V12", "weights", "best.pt"),
    # "Exp13_finetuned": Path(RUNS_DIR, "Test_Run_V13", "weights", "best.pt"),
    # "Exp14_finetuned": Path(RUNS_DIR, "Test_Run_V14", "weights", "best.pt"),
    # "Exp15_finetuned": Path(RUNS_DIR, "Test_Run_V15", "weights", "best.pt"),
    # ... uncomment as each finishes training
}
 
CORAL_QUADRAT2_YAML = Path(CORAL_QUAD_YAML_PATH)
XD38M_YAML = Path(YAML_PATH)
CORALNET_TEST_DIR = Path(CORALNET_TEST_DIR)
 
OUTPUT_DIR = Path(RUNS_DIR, "Grid_Search_Evaluation")
CONF_BANDS = [(0.70, 1.01), (0.40, 0.70), (0.25, 0.40), (0.15, 0.25), (0.05, 0.15), (0.01, 0.05)]
VISUAL_DISPLAY_THRESHOLD = 0.20
EXPORT_VISUALS = True
 
# --- SAHI merge configuration -- tune these if duplicate/nested boxes persist ---
POSTPROCESS_TYPE = "GREEDYNMM"          # better than NMS at catching nested/duplicate boxes
POSTPROCESS_MATCH_METRIC = "IOS"        # intersection-over-smaller-area
POSTPROCESS_MATCH_THRESHOLD = 0.25       # try 0.3 -> 0.2 -> 0.15 if duplicates persist
OVERLAP_RATIO = 0.2                     # try 0.35 if tile-seam duplication persists
SLICE_SIZE = 640
# ===================================================================================
 
# Guard against any two splits silently pointing at the same file.
_yaml_pairs = [("CORAL_QUADRAT2_YAML", CORAL_QUADRAT2_YAML), ("XD38M_YAML", XD38M_YAML)]
for i in range(len(_yaml_pairs)):
    for j in range(i + 1, len(_yaml_pairs)):
        name_a, path_a = _yaml_pairs[i]
        name_b, path_b = _yaml_pairs[j]
        if path_a.resolve() == path_b.resolve():
            raise ValueError(
                f"{name_a} and {name_b} point to the same file:\n  {path_a}\n"
                f"These must be genuinely different datasets -- fix before continuing."
            )
 
_force_postprocess_supported = None  # None = untested, True/False once known
 
 
def sliced_predict_safe(image_path: str, detection_model, **kwargs):
    """get_sliced_prediction wrapper that best-effort tries force_postprocess_type=True
    (some SAHI versions may honor it even on this call path) and transparently falls
    back if the installed version doesn't accept it. Only tests/prints once per run."""
    global _force_postprocess_supported
    if _force_postprocess_supported is not False:
        try:
            result = get_sliced_prediction(image_path, detection_model,
                                            force_postprocess_type=True, **kwargs)
            if _force_postprocess_supported is None:
                print("[info] force_postprocess_type=True accepted by this SAHI version -- using it.")
            _force_postprocess_supported = True
            return result
        except TypeError:
            _force_postprocess_supported = False
            print("[info] This SAHI version's get_sliced_prediction() does not accept "
                  "force_postprocess_type -- continuing without it (the confidence-threshold "
                  "workaround below still applies for the display pass).")
    return get_sliced_prediction(image_path, detection_model, **kwargs)
 
 
def eval_on_split(weights_path: Path, data_yaml: Path, split_label: str) -> dict:
    model = YOLO(str(weights_path))
    metrics = model.val(data=str(data_yaml), split="val", verbose=False)
    return {
        "split": split_label,
        "mAP50": float(metrics.box.map50),
        "mAP50-95": float(metrics.box.map),
        "precision": float(metrics.box.mp),
        "recall": float(metrics.box.mr),
    }
 
 
def confidence_band_counts(confidences):
    counts = {}
    for lo, hi in CONF_BANDS:
        counts[f"{lo:.2f}-{hi:.2f}"] = sum(1 for c in confidences if lo <= c < hi)
    return counts
 
 
def run_sahi_sweep(weights_path: Path, image_dir: Path, label: str, output_dir: Path) -> list:
    rows = []
 
    # Pass 1 model: low floor, for the confidence-band statistics table only.
    stats_model = AutoDetectionModel.from_pretrained(
        model_type="ultralytics", model_path=str(weights_path),
        confidence_threshold=0.01, device="cuda:0",
    )
    # Pass 2 model: real display threshold -- above SAHI's low-confidence cutoff,
    # so GREEDYNMM/IOS merging is honored without needing any workaround.
    display_model = AutoDetectionModel.from_pretrained(
        model_type="ultralytics", model_path=str(weights_path),
        confidence_threshold=VISUAL_DISPLAY_THRESHOLD, device="cuda:0",
    )
 
    image_paths = sorted(p for p in image_dir.glob("*") if p.suffix.lower() in (".jpg", ".jpeg", ".png"))
    if not image_paths:
        print(f"[warn] no images found in {image_dir}")
        return rows
 
    vis_dir = output_dir / "visuals" / label
    if EXPORT_VISUALS:
        vis_dir.mkdir(parents=True, exist_ok=True)
 
    sahi_kwargs = dict(
        slice_height=SLICE_SIZE, slice_width=SLICE_SIZE,
        overlap_height_ratio=OVERLAP_RATIO, overlap_width_ratio=OVERLAP_RATIO,
        postprocess_type=POSTPROCESS_TYPE,
        postprocess_match_metric=POSTPROCESS_MATCH_METRIC,
        postprocess_match_threshold=POSTPROCESS_MATCH_THRESHOLD,
    )
 
    for img_path in image_paths:
        stats_result = sliced_predict_safe(str(img_path), stats_model, **sahi_kwargs)
        confidences = [obj.score.value for obj in stats_result.object_prediction_list]
 
        display_result = sliced_predict_safe(str(img_path), display_model, **sahi_kwargs)
        if EXPORT_VISUALS:
            display_result.export_visuals(export_dir=str(vis_dir), file_name=img_path.stem)
 
        row = {
            "model": label, "image": img_path.name,
            "total_raw_detections": len(confidences),
            "max_confidence": max(confidences) if confidences else 0.0,
            "display_detections_merged": len(display_result.object_prediction_list),
        }
        row.update(confidence_band_counts(confidences))
        rows.append(row)
 
    return rows
 
 
def main():
    OUTPUT_DIR.mkdir(parents=True, exist_ok=True)
    available = {label: path for label, path in EXPERIMENTS.items() if path.exists()}
    missing = set(EXPERIMENTS) - set(available)
    if missing:
        print(f"[warn] skipping {len(missing)} checkpoint(s) not found yet: {sorted(missing)}")
 
    print(f"Evaluating {len(available)} checkpoint(s): {list(available.keys())}\n")
 
    # --- Validation-split comparison ---
    comparison_rows = []
    for label, weights in available.items():
        print(f"[{label}] evaluating on coral-quadrat-2 val...")
        comparison_rows.append({"model": label,
                                 **eval_on_split(weights, CORAL_QUADRAT2_YAML, "coral-quadrat-2 val")})
        print(f"[{label}] evaluating on XD38M val...")
        comparison_rows.append({"model": label,
                                 **eval_on_split(weights, XD38M_YAML, "XD38M val")})
 
    ab_csv = OUTPUT_DIR / "all_experiments_validation_comparison.csv"
    with open(ab_csv, "w", newline="") as f:
        writer = csv.DictWriter(f, fieldnames=["model", "split", "mAP50", "mAP50-95", "precision", "recall"])
        writer.writeheader()
        writer.writerows(comparison_rows)
    print(f"\nSaved: {ab_csv}")
 
    # --- CoralNet confidence sweep (with the SAHI merge fix) ---
    sweep_rows = []
    for label, weights in available.items():
        print(f"\n[{label}] running CoralNet SAHI sweep...")
        sweep_rows += run_sahi_sweep(weights, CORALNET_TEST_DIR, label, OUTPUT_DIR)
 
    if sweep_rows:
        c_csv = OUTPUT_DIR / "all_experiments_coralnet_sweep.csv"
        with open(c_csv, "w", newline="") as f:
            writer = csv.DictWriter(f, fieldnames=list(sweep_rows[0].keys()))
            writer.writeheader()
            writer.writerows(sweep_rows)
        print(f"Saved: {c_csv}")
 
    # --- Consolidated report (no misleading combined score) ---
    lines = []
    lines.append("=" * 100)
    lines.append("EXPERIMENT REPORT -- send this whole file back for diagnosis, nothing else needed")
    lines.append("=" * 100)
 
    splits_present = sorted({r["split"] for r in comparison_rows})
    lines.append("\n--- REAL, LABELED METRICS (per split) ---")
    for split in splits_present:
        is_real_target_domain = "secondary" in split.lower()
        tag = "  <-- closest to real CoralNet domain, weigh most heavily" if is_real_target_domain else ""
        lines.append(f"\n[{split}]{tag}")
        lines.append(f"{'Model':20s} {'Precision':>10s} {'Recall':>10s} {'mAP50':>8s} {'mAP50-95':>10s}")
        for r in comparison_rows:
            if r["split"] == split:
                lines.append(f"{r['model']:20s} {r['precision']:10.3f} {r['recall']:10.3f} "
                              f"{r['mAP50']:8.3f} {r['mAP50-95']:10.3f}")
 
    if sweep_rows:
        lines.append("\n--- CoralNet CONFIDENCE SWEEP (unsupervised proxy -- supporting color only) ---")
        by_model = {}
        for r in sweep_rows:
            by_model.setdefault(r["model"], []).append(r)
        lines.append(f"\n{'Model':20s} {'avg_raw_det':>12s} {'avg_max_conf':>13s} "
                      f"{'avg_hiconf(0.70+)':>18s} {'avg_display_merged':>19s}")
        for model, rs in by_model.items():
            n = len(rs)
            avg_raw = sum(r["total_raw_detections"] for r in rs) / n
            avg_conf = sum(r["max_confidence"] for r in rs) / n
            avg_hi = sum(r["0.70-1.01"] for r in rs) / n
            avg_disp = sum(r["display_detections_merged"] for r in rs) / n
            lines.append(f"{model:20s} {avg_raw:12.1f} {avg_conf:13.2f} {avg_hi:18.1f} {avg_disp:19.1f}")
        lines.append("\n'avg_display_merged' is the corrected, properly-merged detection count at the real")
        lines.append("display threshold -- this is the number to watch for the duplicate-box clutter issue.")
 
    lines.append("\n--- READING GUIDE ---")
    lines.append("1. A real-target-domain split (secondary labeled), if present, decides it -- no further")
    lines.append("   weighing needed.")
    lines.append("2. coral-quadrat-2 / XD38M splits are proxies for two DIFFERENT domains (neither is")
    lines.append("   CoralNet itself) -- a disagreement between them is expected, not a bug.")
    lines.append("3. The confidence-sweep section is context, not a verdict on its own.")
    lines.append("4. No blended score is provided on purpose -- pick one split above and read it directly.")
 
    report = "\n".join(lines)
    print("\n" + report)
 
    report_path = OUTPUT_DIR / "EXPERIMENT_REPORT.txt"
    report_path.write_text(report)
    print(f"\nSaved consolidated report to: {report_path}")
    print(f"All outputs saved to: {OUTPUT_DIR}")
 
 
if __name__ == "__main__":
    main()
```

## CHECKING FOR RESULTS.CSV

```python
from pathlib import Path
import pandas as pd

RUN_NAMES = [
    "Test_Run_V5",
    "Test_Run_V7",
    "Test_Run_V9",
    "Test_Run_V10",
    "modelCalib_Exp1_lowLR_check",
]

MAX_EPOCHS = 80
PATIENCE = 20

rows = []

for run in RUN_NAMES:

    results_path = Path(RUNS_DIR) / run / "results.csv"

    if not results_path.exists():
        print(f"Missing: {results_path}")
        continue

    df = pd.read_csv(results_path)
    df.columns = df.columns.str.strip()

    epoch_col = df.columns[0]

    # Find validation mAP50 column automatically
    map50_col = next(
        c for c in df.columns
        if "mAP50" in c and "95" not in c
    )

    last_epoch = int(df[epoch_col].iloc[-1])
    epochs_completed = last_epoch + 1

    best_idx = df[map50_col].idxmax()
    best_epoch = int(df.loc[best_idx, epoch_col])
    best_map50 = float(df.loc[best_idx, map50_col])

    epochs_after_best = last_epoch - best_epoch

    rows.append({
        "Run": run,
        "Best Epoch": best_epoch,
        "Last Epoch": last_epoch,
        "Epochs Completed": epochs_completed,
        "Best mAP50": round(best_map50, 4),
        "Epochs After Best": epochs_after_best,
        "Reached Max Epoch": epochs_completed == MAX_EPOCHS,
        "Early Stop Likely": (
            epochs_completed < MAX_EPOCHS
            and abs(epochs_after_best - PATIENCE) <= 1
        )
    })

summary = pd.DataFrame(rows)
print(summary.to_string(index=False))
```

## SAHI POST-PROCESSING 

```python
# ---- Cell 0: shared setup ----
# Safe to skip if your directory-setup cell already ran this session; these paths
# just mirror it so this file also works standalone.
 
PROJECT_ROOT = Path.home() / "Documents" / "Thesis"
DATA_DIR = PROJECT_ROOT / "data"
RUNS_DIR = PROJECT_ROOT / "runs"
CORALNET_TEST_DIR = PROJECT_ROOT / "test_images"
CORAL_QUAD_YAML_PATH = DATA_DIR / "coral-quadrat-2-final" / "data.yaml"
 
EXP10_WEIGHTS = RUNS_DIR / "Test_Run_V10" / "weights" / "best.pt"
 
_force_postprocess_supported = None  # same graceful-fallback pattern as Evaluation.py
 
 
def sliced_predict_safe(image_path, detection_model, **kwargs):
    global _force_postprocess_supported
    if _force_postprocess_supported is not False:
        try:
            result = get_sliced_prediction(image_path, detection_model,
                                            force_postprocess_type=True, **kwargs)
            _force_postprocess_supported = True
            return result
        except TypeError:
            _force_postprocess_supported = False
    return get_sliced_prediction(image_path, detection_model, **kwargs)
```

```python
# ---- Cell 1: SAHI post-processing threshold sweep on Exp5 ----
# Your current postprocess_match_threshold=0.5 is not aggressive enough -- every
# Tier 1 visual still shows heavy box overlap after merging. This sweeps 0.5
# (current) down to 0.15 and reports the merged detection count at each, so you
# can see where duplicate clutter actually drops rather than guessing.
#
# Recommendation to start from: 0.25. At 0.5, GREEDYNMM/IOS only merges boxes
# that overlap by more than half their smaller area -- too strict for tile-seam
# duplicates, which are often near-identical but not perfectly overlapping.
 
SLICE_SIZE = 640
OVERLAP_RATIO = 0.2
DISPLAY_CONF_VALUES = [0.15, 0.20]
 
SAHI_MATCH_THRESHOLDS_TO_TEST = [0.50, 0.35, 0.25, 0.15]
 
image_paths = sorted(
    p for p in CORALNET_TEST_DIR.glob("*") if p.suffix.lower() in (".jpg", ".jpeg", ".png")
)
 

for display_conf in DISPLAY_CONF_VALUES:

    display_model = AutoDetectionModel.from_pretrained(
        model_type="ultralytics",
        model_path=str(EXP10_WEIGHTS),
        confidence_threshold=display_conf,
        device="cuda:0",
    )

    print(f"\nDISPLAY_CONF = {display_conf:.2f}")
    print(
        f"{'match_threshold':>16s} "
        f"{'avg_merged':>11s} "
        f"{'min':>5s} "
        f"{'max':>5s}"
    )

    for threshold in SAHI_MATCH_THRESHOLDS_TO_TEST:

        kwargs = dict(
            slice_height=SLICE_SIZE,
            slice_width=SLICE_SIZE,
            overlap_height_ratio=OVERLAP_RATIO,
            overlap_width_ratio=OVERLAP_RATIO,
            postprocess_type="GREEDYNMM",
            postprocess_match_metric="IOS",
            postprocess_match_threshold=threshold,
        )

        merged_counts = []

        for img_path in image_paths:

            result = sliced_predict_safe(
                str(img_path),
                display_model,
                **kwargs
            )

            merged_counts.append(
                len(result.object_prediction_list)
            )

        avg = sum(merged_counts) / len(merged_counts)

        print(
            f"{threshold:16.2f} "
            f"{avg:11.1f} "
            f"{min(merged_counts):5d} "
            f"{max(merged_counts):5d}"
        )
 
# To visually confirm whichever threshold looks best numerically, re-run with
# EXPORT_VISUALS-style export_visuals(...) on that one config -- see Evaluation.py's
# run_sahi_sweep for the export call if you want that added back in.
```

```python
# ---- Cell 2: data-driven confidence threshold from labeled coral-quadrat-2 val ----
# The CoralNet sweep has no ground truth, so it can tell you about clutter but not
# about the right confidence cutoff. coral-quadrat-2 val DOES have ground truth --
# this sweeps conf there and reports precision/recall/F1 per model, so the choice
# is grounded in labels instead of eyeballing the unlabeled CoralNet images.
#
# Exp8 is intentionally excluded until the Test_Run_V7 / Test_Run_V8 path mix-up
# in Evaluation.py's EXPERIMENTS dict is fixed (see the diagnosis) and Exp8 is
# re-evaluated against its own checkpoint.
 
CANDIDATE_WEIGHTS = {
    "Exp5_finetuned": RUNS_DIR / "Test_Run_V5" / "weights" / "best.pt",
    "Exp7_finetuned": RUNS_DIR / "Test_Run_V7" / "weights" / "best.pt",
    "Exp9_finetuned": RUNS_DIR / "Test_Run_V9" / "weights" / "best.pt",
    "Exp10_finetuned": RUNS_DIR / "Test_Run_V10" / "weights" / "best.pt",
}
 
CONF_CANDIDATES = [0.10, 0.15, 0.20, 0.25, 0.30, 0.35, 0.40]
 
for label, weights in CANDIDATE_WEIGHTS.items():
    if not weights.exists():
        print(f"[skip] {label}: weights not found at {weights}")
        continue
    model = YOLO(str(weights))
    print(f"\n{label}")
    print(f"  {'conf':>5s} {'precision':>10s} {'recall':>8s} {'f1':>6s}")
    best_f1, best_conf = -1.0, None
    for conf in CONF_CANDIDATES:
        metrics = model.val(data=str(CORAL_QUAD_YAML_PATH), split="val", conf=conf, verbose=False)
        p, r = float(metrics.box.mp), float(metrics.box.mr)
        f1 = 2 * p * r / (p + r) if (p + r) > 0 else 0.0
        print(f"  {conf:5.2f} {p:10.3f} {r:8.3f} {f1:6.3f}")
        if f1 > best_f1:
            best_f1, best_conf = f1, conf
    print(f"  -> best F1 at conf={best_conf} (F1={best_f1:.3f})")
```
